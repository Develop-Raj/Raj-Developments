```python
import numpy as np
```

# Learing the Basics

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([10,12,14,16,15])

print(arr1+arr2)
```

```python
multi = np.array([[1,2,3],[1,2,3],[1,2,3]])
multi
```

## some basics functions

```python
arr = np.ones(1);
print(arr);

arr = np.ones((5,2))
print(arr)
```

```python
arr = np.zeros(5)
print(arr)
arr = np.zeros((4,4))
print(arr)
```

```python
arr = np.arange(1,11)
print(arr)
```

```python
arr = np.arange(1,11,3)
arr
```

```python
arr = np.linspace(0,10,1)
print(arr)

arr = np.linspace(0,10,2)
print(arr)

arr = np.linspace(0,10,5)
print(arr)
```

```python
arr = np.eye(5)
arr
```
