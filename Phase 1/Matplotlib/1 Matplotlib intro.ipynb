```python
import matplotlib.pyplot as m
import numpy as np
```

```python
x = np.array([2025,2026,2027])
y = np.array([100, 70, 150])


m.plot(x,y)
m.show
```

```python
m.plot(y)
```

```python
m.plot(x)
```

```python
m.plot(x,marker=".",markersize=30,
       markerfacecolor="red",
       markeredgecolor="red",
       linestyle="dashed")
```

```python
m.plot(x,marker=".",markersize=30,
       markerfacecolor="red",
       markeredgecolor="red",
       linestyle="None")
```

```python
m.plot(x,marker=".",markersize=30,
       markerfacecolor="red",
       markeredgecolor="red",
       linestyle="dotted",
       linewidth=10,
       color="grey")
```

```python
plotDesign = dict(marker=".",
                markersize=30,
                markerfacecolor="red",
                markeredgecolor="red",
                linestyle="solid",
                linewidth=5)

y1 = np.array([90,70,80,77,93])
y2 = np.array([50,65,73,70,83])
y3 = np.array([30,59,57,63,97])

m.plot(y1,color="blue",**plotDesign)
m.plot(y2,color="green",**plotDesign)
m.plot(y3,**plotDesign)
```
