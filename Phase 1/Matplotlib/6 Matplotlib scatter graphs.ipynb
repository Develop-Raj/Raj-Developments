```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([0,10,20,10,15,12,13,13])
y = np.array([55,68,99,63,91,77,78,84])

plt.xlabel("Hours Studied")
plt.ylabel("Marks obtained")
plt.scatter(x,y,color="black")
plt.show()
```

```python
plt.scatter(x,y,color="black",alpha=0.4, s=100)
```

```python
x1 = np.array([0,2,3,5])
y1 = np.array([55,68,99,63])

x2 = np.array([1,2.5,3,5])
y2 = np.array([60,80,77,95])

plt.scatter(x1,y1,color="black", alpha=0.3,label="Class A")
plt.scatter(x2,y2,color="red",label="Class B")
plt.legend()
plt.show()
```
