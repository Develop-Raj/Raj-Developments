```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
foods = np.array(["Grains","Fruits","Protein","Dairy","Vegetables"])
values = np.array([4,3,2,5,3])

colors = ["yellow","red","blue","grey","green"]
plt.title("Foods Consumption")
plt.pie(values,labels = foods, autopct = "%1.1f",colors=colors)
plt.show()
```
