```python
import matplotlib.pyplot as pl
import numpy as np
```

```python
x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,25])
y2 = np.array([17,24,28,19])
```

```python
pl.title("Class Size", fontsize=20,family="Arial",color="#2d4cfc")


pl.plot(x,y1)
pl.plot(x,y2)

pl.show()
```

```python
pl.xlabel("Year",fontsize=20)
pl.ylabel("Students",fontsize=20)
pl.title("Class Size", fontsize=20,family="Arial",color="#2d4cfc")


pl.plot(x,y1)
pl.plot(x,y2)

pl.show()
pl.show()
```

```python
pl.xlabel("Year",fontsize=20)
pl.ylabel("Students",fontsize=20)
pl.title("Class Size", fontsize=20,family="Arial",color="#2d4cfc")


pl.plot(x,y1)
pl.plot(x,y2)
pl.xticks(x)
pl.show()
```
