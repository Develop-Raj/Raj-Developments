```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
foods = np.array(["Grains","Fruits","Protein","Dairy","Vegetables"])
values = np.array([4,3,2,5,3])

plt.bar(foods,values)
plt.show()
```

```python
foods = np.array(["Grains","Fruits","Protein","Dairy","Vegetables"])
values = np.array([4,3,2,5,3])
plt.title("Foods Consumption")
plt.xlabel("Foods")
plt.ylabel("Quantity")
plt.bar(foods,values,color="#fc237c")
plt.show()
```

```python
foods = np.array(["Grains","Fruits","Protein","Dairy","Vegetables"])
values = np.array([4,3,2,5,3])
plt.title("Foods Consumption")
plt.xlabel("Foods")
plt.ylabel("Quantity")
plt.barh(foods,values,color="#fc237c")
plt.show()
```
