```python
import pandas as pd
```

```python
# panel data - pandas
print(pd.__version__) # checking the version


data = [100,102,104]

series = pd.Series(data)

print(series)
```

```python

data = [100.2,102.4,104.6] # series can of different types float, int, string, boolean

series = pd.Series(data)

print(series)
```

```python
data = [100,102,104]

series = pd.Series(data,index = ["a","b","c"])

series
```

```python
print(series.loc["a"])
print(series.loc["b"])
print(series.loc["c"])

series.loc["c"] = 200

print(series)

print(series.iloc[0])
print(series.iloc[1])
print(series.iloc[2])
```

```python
data = [100,101,102,103,104,105]

series = pd.Series(data,index=["A","B","C","D",'E','F'])
print(series)

print(series[series >= 103])
print(series[series < 103])
```

```python
calories = {"day 1": 1750,"day 2": 1800,"day 3": 1700}

series = pd.Series(calories)
series
```

```python
print(series.loc["day 1"])
print(series.loc["day 2"])

series.loc["day 3"] += 500
print(series.loc["day 3"])
```
