```python
import pandas as pd
```

```python
data = {"Name":["Raj","Sabina","Lily","Rose"],
        "Age":[20,20,0,0]}
df = pd.DataFrame(data,index=["Father","Mother","Daughter 1","Daughter 2"])
print(df)
print(df.loc["Father"])
print(df.iloc[0])
```

```python
# add a new Column
df["Job"] = ["Engineer", "Dentist", "N/A", "N/a"]

print(df)
```

```python
new_row = pd.DataFrame([{"Name":"Son 1", "Age" : 0, "Job" : "N/A"}])

df = pd.concat([df,new_row])

df
```

```python
new_row = pd.DataFrame([{"Name":"Son 1", "Age" : 0, "Job" : "N/A"}],index=["Son"])

df = pd.concat([df,new_row])

df
```
