```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")

print(df.mean(numeric_only=True))
print(df.sum(numeric_only=True))
print(df.min(numeric_only=True))
print(df.max(numeric_only=True))
```

```python
print(df.count())
```

```python
print(df["Age"].mean())
print(df["Age"].sum())
print(df["Age"].min())
print(df["Age"].max())
```

```python
group = df.groupby("Gender")
print(group["Science"].mean())
print(group["Age"].mean())
print(group["Math"].mean())
print(group["English"].mean())
```
